# Question 2 part 3

From question 1 part 2 we know that the Cartesian form of the normalized quantum state is:

$\ket{\psi_{norm}} = \sqrt{\frac{0.7}{1.2}}\ket{0} + (\frac{0.5}{\sqrt{1.2}}+\frac{0.5}{\sqrt{1.2}}i)\ket{1}$

Polar form:

$\ket{\psi_{norm}} = \cos{\frac{\theta}{2}}\ket{0} + e^{{i}{\phi}}\sin{\frac{\theta}{2}}\ket{1}$

Such that $\theta \approx 1.4$ , $\phi = \frac{\pi}{4}$

Which can be re-written as:

$\ket{\psi_{norm}} = \sqrt{\frac{7}{12}}\ket{0} + \sqrt{\frac{5}{12}}e^{{i}{\frac{\pi}{4}}}\ket{1}$

From it we constructed the Bloch vector to be:

$\vec{r} = (\frac{\sqrt{0.7}}{1.2},\frac{\sqrt{0.7}}{1.2},\frac{1}{6})$

## Constructing normalized state with qiskit

````python
import numpy as np 
from qiskit.quantum_info import Statevector
from qiskit import QuantumCircuit

alpha = np.sqrt(7/12)
beta = np.sqrt(5/12)*np.exp(1j * (np.pi/4))

sv = Statevector([alpha,beta])

print(sv)

qc = QuantumCircuit(1)
qc.prepare_state(sv,0)
````

## Verifying results with qiskit

We will use ```plot_bloch_multivector``` to plot our statevector, and ```plot_bloch_vector``` to plot our values of $x,y,z$ we found in part 2 of question 3. If both Bloch spheres are identical then our result is correct. 

1. Plotting statvector $\ket{\psi_{norm}} = \sqrt{\frac{7}{12}}\ket{0} + \sqrt{\frac{5}{12}}e^{{i}{\frac{\pi}{4}}}\ket{1}$

````python
from qiskit.visualization import plot_bloch_vector
from qiskit.visualization import plot_bloch_multivector

plot_bloch_multivector(sv)
````

2. Plotting $\vec{r} = (\frac{\sqrt{0.7}}{1.2},\frac{\sqrt{0.7}}{1.2},\frac{1}{6})$

````python
plot_bloch_vector([np.sqrt(0.7)/1.2,np.sqrt(0.7)/1.2,1/6])
````

Both Bloch sphere representations are identical therefore our solution is correct.
